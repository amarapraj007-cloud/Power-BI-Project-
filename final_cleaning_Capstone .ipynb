# Handle Missing Values

### For Score Columns → optional median fill

```python
score_cols = [col for col in df.columns if 'Score' in col]

df[score_cols] = df[score_cols].fillna(df[score_cols].median())
```

### For Rank Columns

In this dataset, missing values (NaN) are present in several ranking-related columns such as Academic Reputation Rank, Employer Reputation Rank, and other indicator-specific ranks. These missing values do not indicate poor performance or a low ranking. Instead, they represent cases where a university was not evaluated or not ranked for a specific indicator.

Since ranking variables are ordinal by nature, imputing or replacing missing values with numerical placeholders (such as 0, maximum rank, or average rank) would introduce incorrect assumptions and distort the relative ordering of institutions. Similarly, removing records with missing rank values would lead to unnecessary data loss and bias country-level and institution-level analysis.

Therefore, missing values in ranking columns were intentionally retained as NaN to preserve the integrity and meaning of the data. During analysis and visualization, these NaN values are interpreted as “Not Ranked” and are handled at the presentation layer (e.g., tooltips, filters, or explanatory notes) rather than being altered in the underlying data.

This approach ensures:

Accurate statistical calculations

Correct handling of numeric data types

Transparent and unbiased analytical insights
